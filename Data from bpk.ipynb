# One item page

```python
def get_data(url):
    datum = {}
    response = requests.get(url, headers=headers)
    soup = BeautifulSoup(response.content, "html.parser")
    containers = soup.select("div.card.mb-6")
    metadata = containers[0].select_one("div.container.fs-6")
    
    download_tag = containers[1].find(class_='download-file')
    download_url = base_url + download_tag['href']
    datum['download_url'] = download_url
    
    for div in metadata.find_all("div", recursive=False):
        text = div.get_text(" ", strip=True)
    
        text = text.upper()
    
        if "TIPE DOKUMEN" in text:
            value = text.split('DOKUMEN')[1].strip()
            datum['tipe_dokumen'] = doc_type
    
        elif "JUDUL" in text:
            value = text.split('JUDUL')[1].strip()
            datum['judul'] = value
    
        elif "T.E.U." in text:
            value = text.split('T.E.U.')[1].strip()
            datum['teu'] = value
    
        elif "BENTUK SINGKAT" in text:
            value = text.split('SINGKAT')[1].strip()
            datum['bentuk_singkat'] = value
    
        elif "BENTUK" in text:
            value = text.split('BENTUK')[1].strip()
            datum['bentuk'] = value
    
        elif "ALIAS" in text:
            value = text.split('ALIAS')[1].strip()
            datum['alias'] = value
    
        elif "TAHUN" in text and "NOMOR" not in text:
            value = text.split('TAHUN')[1].strip()
            datum['tahun'] = value
    
        elif "TEMPAT PENETAPAN" in text:
            value = text.split('PENETAPAN')[1].strip()
            datum['tempat_penetapan'] = value
    
        elif "TANGGAL PENETAPAN" in text:
            value = text.split('PENETAPAN')[1].strip()
            datum['tanggal_penetapan'] = value
    
        elif "TANGGAL PENGUNDANGAN" in text:
            value = text.split('PENGUNDANGAN')[1].strip()
            datum['tanggal_pengundangan'] = value
    
        elif "TANGGAL BERLAKU" in text:
            value = text.split('BERLAKU')[1].strip()
            datum['tanggal_berlaku'] = value
    
        elif "SUMBER" in text:
            value = text.split('SUMBER')[1].strip()
            datum['sumber'] = value
    
        elif "SUBJEK" in text:
            value = text.split('SUBJEK')[1].strip()
            datum['subjek'] = value
    
        elif "STATUS" in text:
            value = text.split('STATUS')[1].strip()
            datum['status'] = value
    
        elif "LOKASI" in text:
            value = text.split('LOKASI')[1].strip()
            datum['lokasi'] = value
    
        elif "BIDANG" in text:
            value = text.split('BIDANG')[1].strip()
            datum['bidang'] = value
    
        elif "NOMOR" in text:
            value = text.split('NOMOR')[1].strip()
            datum['nomor'] = value
    


    status_peraturan = containers[2].select_one("div.container.fs-6")
    current_key = None
    if status_peraturan:
        for div in status_peraturan.find_all("div", recursive=False):
            text = div.get_text(" ", strip=True)
            text_uppercase = text.upper()
        
            if current_key is None:
                if "DICABUT DENGAN" in text_uppercase:
                    current_key = "dicabut_dengan"
        
                elif "DICABUT SEBAGIAN" in text_uppercase:
                    current_key = 'dicabut_sebagian_dengan'
        
                elif "DIUBAH DENGAN" in text_uppercase:
                    current_key = 'diubah_dengan'
        
                elif "MENCABUT" in text_uppercase:
                    current_key = 'mencabut'
        
                elif "MENGUBAH" in text_uppercase:
                    current_key = 'mengubah'
            
            else:
                datum[current_key] = text
                current_key = None

    return datum
```
